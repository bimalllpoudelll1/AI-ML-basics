```python
# feature selection is a technique in which we select columns from the data that a important and also
# figure out and remove those which are less important than others for data sciencee and ML
# there are two types, forward and backward eleminaion to train model with highest accuracy
import pandas as pd
from mlxtend.feature_selection import SequentialFeatureSelector
dataset=pd.read_csv(r"C:\Users\dell\Documents\excells\data1.csv")
dataset.head(3)
```

```python
dataset.shape
```

```python
x=dataset.iloc[:,:-1] #iloc is integers location
y=dataset["Propertycount"]
```

now we need to make a model 

```python
x.info()
```

```python
x.shape
```

```python
x.isnull().sum()
```

```python
num=x.select_dtypes(include=['int64','float64']).columns
x[num]=x[num].fillna(x[num].median())
```

```python
cat_data=x.select_dtypes(include=['object']).columns
x[cat_data]=x[cat_data].fillna(x[cat_data].mode().iloc[0])
```

```python
x.isnull().sum()
```

```python
x=pd.get_dummies(x,drop_first=True)
x
```

```python
x.shape
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
FS=SequentialFeatureSelector(lr, k_features=2, forward=True, scoring='r2',cv=2, n_jobs=-1)
FS.fit(x,y)
```
