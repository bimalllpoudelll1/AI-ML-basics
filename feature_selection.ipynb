```python
# feature selection is a technique in which we select columns from the data that a important and also
# figure out and remove those which are less important than others for data sciencee and ML
# there are two types, forward and backward eleminaion to train model with highest accuracy
import pandas as pd
from mlxtend.feature_selection import SequentialFeatureSelector
dataset=pd.read_csv(r"ONEHOTENCOE.csv")
dataset
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python
dataset.info()
```

```python
num=dataset.select_dtypes(include=['int64','float64']).columns
dataset[num]=dataset[num].fillna(dataset[num].median())
```

```python
cat_data=dataset.select_dtypes(include=['object']).columns
dataset[cat_data]=dataset[cat_data].fillna(dataset[cat_data].mode().iloc[0])
```

```python
dataset
```

```python
# one hot encoding
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
x=dataset.iloc[:,:-1] #iloc is integers location
y=dataset["Rating"]
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
FS=SequentialFeatureSelector(lr, k_features=2, forward=True)
FS.fit(x,y)
```
