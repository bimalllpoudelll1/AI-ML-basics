```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
dataset=pd.read_csv(r"C:\Users\dell\Documents\mine server\cgpa_package_500.csv")
dataset.head(3)
```

```python
# we using scater plot
sns.scatterplot(x="cgpa",y="package",data=dataset)
plt.show()
```

```python
dataset.isnull().sum()
```

```python
x=dataset[["cgpa"]]
y=dataset["package"]
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# sklearn is a library in which linear model is a module in that modeule there is a clas sname LinearRegression
# now we convert the clss into object as we cannot work withot any object
from sklearn.linear_model import LinearRegression 
lr=LinearRegression()
lr.fit (x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
# predciting the accuracy of the model
lr.predict([[5.00]])
```

```python
# for linear regression we use y=mx+c
# m is the coefficent and c is the intercept
print(lr.coef_)
print(lr.intercept_)
```
