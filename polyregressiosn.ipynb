```python
'''
we use polynomila regression when the graph doesnt follow linerrity and has curve nature.
y=a1​x+a2​x^2+a3​x^3+⋯+an​x^n+c
it has power whereas multiple regression has none
'''
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
dataset=pd.read_csv("datas\polyregress.csv")
dataset.head(4)
```

```python
dataset.corr()
```

```python
plt.scatter(dataset['Level'] , dataset['Salary'])
plt.xlabel('Level')
plt.ylabel('Salary')
plt.show()
```

```python
'''
now converting into polynomill feature
we seperate the dependent and independett variable

'''
# seperating x and y
x=dataset[['Level']]
y=dataset['Salary']
```

```python
from sklearn.preprocessing import PolynomialFeatures
PF=PolynomialFeatures(degree=3) #we using degree 3 as 2 wasnt fitting perfectly
PF.fit(x)
x=PF.transform(x)
```

```python
# after polyynomial feautyre we split into train test
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test= train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train, y_train)
```

```python
# now we check the accuracy
lr.score(x_test,y_test)*100
```

```python
pdr=lr.predict(x)
```

```python
plt.scatter(dataset['Level'],dataset['Salary'])
plt.plot(dataset['Level'],pdr,c='red')
plt.xlabel('Lavel')
plt.ylabel('Salary')
plt.legend(["org",'prd'])
plt.show()
```
